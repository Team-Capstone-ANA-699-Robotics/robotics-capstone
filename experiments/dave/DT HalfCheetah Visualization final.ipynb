## Section 2: Model Architecture

The `DecisionTransformer` class is reproduced here verbatim from the training notebook.  
This is required because PyTorch's `torch.load` restores weights into an instantiated model object — the class definition must be present in scope before loading the checkpoint.

> **Important:** If you modify the architecture in the training notebook, you must update this cell to match. A mismatch between the saved weights and this definition will raise a `RuntimeError` on `load_state_dict`.

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 2: Decision Transformer Architecture
#
# Mirrored exactly from the training notebook (Section 3).
# GPT-2 backbone with three separate linear embedding heads:
#   - state embedding
#   - action embedding
#   - return-to-go (RTG) embedding
# A final linear layer projects the transformer output to action space.
# ─────────────────────────────────────────────────────────────────────────────

class DecisionTransformer(nn.Module):
    """
    Decision Transformer — architecture matched exactly to the saved checkpoint.

    Key differences from the placeholder in the original visualization notebook:
      - Embedding layers use flat Linear (no Sequential+LayerNorm wrapper):
            embed_rtg, embed_state, embed_action
      - A single shared LayerNorm (embed_ln) is applied AFTER summing all three
        embedded tokens, rather than per-modality LayerNorms
      - Timestep embedding is named embed_timestep (shape: 1000 x 128)
      - GPT-2 is configured with n_ctx=90 (= 3 * context_len = 3 * 30)
        and vocab_size=1 (token embeddings are bypassed — inputs_embeds is used)
      - Output head is named predict_action (single Linear + Tanh, no Sequential)
    """

    def __init__(
        self,
        state_dim   : int,
        action_dim  : int,
        hidden_dim  : int,
        context_len : int,
        n_layers    : int,
        n_heads     : int,
        dropout     : float,
        max_ep_len  : int = 1000,
    ):
        super().__init__()

        self.state_dim   = state_dim
        self.action_dim  = action_dim
        self.hidden_dim  = hidden_dim
        self.context_len = context_len

        # ── Timestep embedding ────────────────────────────────────────────────
        # Shape matches checkpoint: (max_ep_len=1000, hidden_dim=128)
        self.embed_timestep = nn.Embedding(max_ep_len, hidden_dim)

        # ── Input projection heads (flat Linear, no LayerNorm wrapper) ────────
        # Each modality projected independently to hidden_dim.
        self.embed_rtg    = nn.Linear(1,          hidden_dim)
        self.embed_state  = nn.Linear(state_dim,  hidden_dim)
        self.embed_action = nn.Linear(action_dim, hidden_dim)

        # ── Shared LayerNorm applied after summing embeddings ─────────────────
        # Applied once to the combined token embedding, not per-modality.
        self.embed_ln = nn.LayerNorm(hidden_dim)

        # ── GPT-2 backbone ────────────────────────────────────────────────────
        # vocab_size=1  : token embeddings (wte) are unused — we pass
        #                 inputs_embeds directly, but GPT2Model still
        #                 instantiates wte; size=1 keeps it minimal.
        # n_positions=90: matches checkpoint wpe shape (90, 128) = 3 * context_len
        gpt2_cfg = GPT2Config(
            vocab_size    = 1,
            n_embd        = hidden_dim,
            n_layer       = n_layers,
            n_head        = n_heads,
            n_positions   = 3 * context_len,
            n_ctx         = 3 * context_len,
            resid_pdrop   = dropout,
            embd_pdrop    = dropout,
            attn_pdrop    = dropout,
            use_cache     = False,
        )
        self.transformer = GPT2Model(gpt2_cfg)

        # ── Output head ───────────────────────────────────────────────────────
        # Single Linear layer + Tanh, stored as a one-element nn.ModuleList
        # to match the checkpoint key: predict_action.0.weight / .bias
        self.predict_action = nn.ModuleList([nn.Linear(hidden_dim, action_dim)])

    def forward(
        self,
        states    : torch.Tensor,   # (B, K, state_dim)
        actions   : torch.Tensor,   # (B, K, action_dim)
        rtg       : torch.Tensor,   # (B, K, 1)
        timesteps : torch.Tensor,   # (B, K)
        mask      : torch.Tensor,   # (B, K) or None
    ) -> torch.Tensor:              # (B, K, action_dim)

        B, K, _ = states.shape

        # ── Timestep encoding (shared across all three token types) ───────────
        t_emb = self.embed_timestep(timesteps)   # (B, K, H)

        # ── Project each modality and add timestep encoding, then LayerNorm ───
        r_tok = self.embed_ln(self.embed_rtg(rtg)       + t_emb)
        s_tok = self.embed_ln(self.embed_state(states)  + t_emb)
        a_tok = self.embed_ln(self.embed_action(actions)+ t_emb)

        # ── Interleave into [r0, s0, a0, r1, s1, a1, ...] ────────────────────
        seq = torch.stack([r_tok, s_tok, a_tok], dim=2)  # (B, K, 3, H)
        seq = seq.reshape(B, 3 * K, self.hidden_dim)     # (B, 3K, H)

        # ── Attention mask ────────────────────────────────────────────────────
        if mask is not None:
            attn_mask = mask.unsqueeze(-1).repeat(1, 1, 3).reshape(B, 3 * K)
        else:
            attn_mask = torch.ones(B, 3 * K, device=states.device)

        # ── GPT-2 forward pass ────────────────────────────────────────────────
        out = self.transformer(
            inputs_embeds  = seq,
            attention_mask = attn_mask,
        ).last_hidden_state   # (B, 3K, H)

        # ── Extract state-token positions and predict actions ─────────────────
        # State tokens sit at positions 1, 4, 7, ... (offset 1, stride 3)
        state_out = out[:, 1::3, :]                          # (B, K, H)
        action_preds = torch.tanh(self.predict_action[0](state_out))  # (B, K, action_dim)

        return action_preds


print("DecisionTransformer class defined (matched to checkpoint).")
````

## Section 3: Load Checkpoint & Normalization Stats

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 3.1: Load the trained model checkpoint
# ─────────────────────────────────────────────────────────────────────────────

# Verify the checkpoint file exists before attempting to load
ckpt_path = Path(CFG.checkpoint_path)
assert ckpt_path.exists(), (
    f"Checkpoint not found at {ckpt_path}.\n"
    f"Ensure the training notebook completed its full run and saved to ./checkpoints/."
)

# Load the checkpoint dict from disk.
# map_location ensures the tensor device matches our current DEVICE,
# which is important when the checkpoint was saved on MPS and reloaded on CPU
# (or vice versa, e.g. when moving between machines).
checkpoint = torch.load(ckpt_path, map_location=DEVICE)

print(f"Checkpoint loaded from: {ckpt_path}")
print(f"  Keys in checkpoint : {list(checkpoint.keys())}")
if 'iteration' in checkpoint:
    print(f"  Saved at iteration : {checkpoint['iteration']:,}")
if 'loss' in checkpoint:
    print(f"  Final loss         : {checkpoint['loss']:.6f}")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 3.2: Instantiate the model and restore weights
# ─────────────────────────────────────────────────────────────────────────────

# Instantiate a fresh DecisionTransformer using the same hyperparameters
# as the training run (defined in CFG). This creates an empty model shell
# whose architecture exactly matches the saved weights.
model = DecisionTransformer(
    state_dim   = CFG.state_dim,
    action_dim  = CFG.action_dim,
    hidden_dim  = CFG.hidden_dim,
    context_len = CFG.context_len,
    n_layers    = CFG.n_layers,
    n_heads     = CFG.n_heads,
    dropout     = CFG.dropout,
    max_ep_len  = CFG.max_ep_len,
).to(DEVICE)

# Restore the saved weights into the model.
# strict=True (default) ensures every key in the checkpoint matches the model
# exactly — any mismatch raises an error rather than silently loading partial weights.
model.load_state_dict(checkpoint['model_state'])

# Switch the model to inference mode:
#   - disables dropout
#   - disables gradient tracking (no backprop needed)
model.eval()

total_params = sum(p.numel() for p in model.parameters())
print(f"Model restored successfully.")
print(f"  Total parameters : {total_params:,}")
print(f"  Device           : {next(model.parameters()).device}")
print(f"  Mode             : {'eval (inference)'}")
````

## Section 4: Rendering Utilities

Two utilities are defined here:

1. **`rollout_and_record`** — runs a single MuJoCo episode using the DT policy, captures every rendered frame, writes an MP4 to disk via `imageio`, and returns the episode return.
2. **`display_video`** — reads a saved MP4, base64-encodes it, and injects it as an HTML5 `<video>` element for inline notebook display.

> **Why `rgb_array` instead of `human` render mode?**  
> `render_mode='human'` opens a native OS window and conflicts with Jupyter's event loop on macOS, often causing kernel hangs. `render_mode='rgb_array'` renders each frame to a numpy array silently — no window is opened. `imageio` then assembles the frames into an MP4 using its bundled ffmpeg binary.

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4.1: Rollout + recording function
# ─────────────────────────────────────────────────────────────────────────────

@torch.no_grad()  # disables gradient computation for the entire rollout
def rollout_and_record(
    model       : DecisionTransformer,
    env         : gym.Env,
    target_rtg  : float,
    state_mean  : np.ndarray,
    state_std   : np.ndarray,
    device      : torch.device,
    context_len : int,
    rtg_scale   : float,
    video_path  : str,
    fps         : int = 30,
    max_ep_len  : int = 1000,
) -> Tuple[float, List[float]]:
    """
    Run one MuJoCo episode with the trained DT policy, record every frame,
    and save the result as an MP4 at `video_path`.

    Returns:
        total_return   (float)      : sum of undiscounted rewards over the episode
        reward_history (List[float]): per-step rewards for post-hoc analysis
    """
    model.eval()
    frames = []   # accumulate rendered frames as numpy arrays (H, W, 3)

    # ── Initialize episode ────────────────────────────────────────────────────
    state, _ = env.reset()
    # Normalize the initial observation using dataset statistics
    state = (state - state_mean) / state_std

    # ── Rolling context buffers (K = context_len timesteps) ───────────────────
    # The DT is auto-regressive over a sliding window of K steps.
    # We maintain fixed-size buffers and roll them left at each step.
    states_buf    = torch.zeros((1, context_len, CFG.state_dim),  dtype=torch.float32, device=device)
    actions_buf   = torch.zeros((1, context_len, CFG.action_dim), dtype=torch.float32, device=device)
    rtg_buf       = torch.zeros((1, context_len, 1),              dtype=torch.float32, device=device)
    timesteps_buf = torch.zeros((1, context_len),                 dtype=torch.long,    device=device)

    current_rtg   = target_rtg   # decremented by the observed reward at each step
    total_return  = 0.0
    reward_history = []

    for t in range(max_ep_len):

        # ── Capture the rendered frame ─────────────────────────────────────
        # env.render() returns an (H, W, 3) uint8 numpy array in rgb_array mode.
        frame = env.render()
        if frame is not None:
            frames.append(frame)



        # ── Shift context buffers left, insert new observation at the end ──
        # torch.roll shifts elements along a dimension; the rightmost slot (-1)
        # always holds the most recent timestep's data.
        states_buf    = torch.roll(states_buf,    -1, dims=1)
        rtg_buf       = torch.roll(rtg_buf,       -1, dims=1)
        timesteps_buf = torch.roll(timesteps_buf, -1, dims=1)

        states_buf[0, -1, :]  = torch.tensor(state,           dtype=torch.float32, device=device)
        rtg_buf[0, -1, 0]     = current_rtg / rtg_scale       # normalize RTG to match training scale
        timesteps_buf[0, -1]  = t

        # ── Forward pass: predict action for the current context ───────────
        # The model receives the full K-step context and returns action
        # predictions for all K positions; we take only the last one.
        action_pred = model(
            states    = states_buf,
            actions   = actions_buf,
            rtg       = rtg_buf,
            timesteps = timesteps_buf,
            mask      = None,   # no padding mask needed during inference
        )
        action = action_pred[0, -1, :].cpu().numpy()   # shape: (action_dim,)

        # ── Insert the predicted action into the action buffer ─────────────
        actions_buf = torch.roll(actions_buf, -1, dims=1)
        actions_buf[0, -1, :] = torch.tensor(action, dtype=torch.float32, device=device)

        # ── Step the environment ───────────────────────────────────────────
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        # ── Update RTG: decrement by the reward just observed ─────────────
        # This is the core DT inference loop:
        #   RTG_{t+1} = RTG_t - r_t
        # The agent is always conditioned on "how much return is still needed".
        current_rtg  -= reward
        total_return += reward
        reward_history.append(reward)

        # Normalize next observation before inserting into context
        state = (next_state - state_mean) / state_std

        if done:
            break

    # ── Write frames to MP4 ───────────────────────────────────────────────────
    # imageio uses its bundled ffmpeg binary (from imageio-ffmpeg) to encode
    # the frame list as an MP4. The 'ffmpeg' plugin is the most reliable on macOS.
    if frames:
        imageio.mimwrite(
            video_path,
            frames,
            fps=fps,
            codec='libx264',        # H.264 codec — broadly compatible
            quality=8,              # 1 (lowest) – 10 (highest); 8 is a good balance
            macro_block_size=None,  # avoids ffmpeg dimension rounding issues
        )

    return total_return, reward_history


print("rollout_and_record() defined.")
````

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4.2: Inline HTML5 video display helper
# ─────────────────────────────────────────────────────────────────────────────

def display_video(
    video_path   : str,
    episode_idx  : int,
    total_return : float,
    norm_return  : float,
    width        : int = 480,
) -> None:
    """
    Read a saved MP4 file, base64-encode it, and render it as an HTML5
    <video> element directly in the Jupyter notebook output cell.

    This approach works without a local HTTP server and is fully portable —
    the video data is embedded in the notebook's cell output.

    Args:
        video_path   : path to the saved MP4 file
        episode_idx  : episode number (for the display label)
        total_return : raw undiscounted episode return
        norm_return  : D4RL-normalized return (0–100 scale)
        width        : display width in pixels (height scales proportionally)
    """
    # Read the raw MP4 bytes and base64-encode them for embedding in HTML
    with open(video_path, 'rb') as f:
        video_bytes = f.read()
    b64 = base64.b64encode(video_bytes).decode('utf-8')

    # Build an HTML5 video tag with autoplay and loop for smooth demo viewing.
    # The data URI embeds the full MP4 — no separate file server needed.
    html = f"""
    <div style="margin: 12px 0; font-family: monospace;">
      <b>Episode {episode_idx + 1:02d}</b> &nbsp;|
      Raw return: <b>{total_return:.1f}</b> &nbsp;|
      D4RL normalized: <b>{norm_return:.1f}</b>
      <br/>
      <video width="{width}" controls autoplay loop muted
             style="margin-top: 6px; border: 1px solid #ccc; border-radius: 4px;">
        <source src="data:video/mp4;base64,{b64}" type="video/mp4">
        Your browser does not support the HTML5 video tag.
      </video>
    </div>
    """
    display(HTML(html))


def normalize_return(raw: float, d4rl_random: float, d4rl_expert: float) -> float:
    """
    Convert a raw episode return to the D4RL-normalized score (0–100 scale).
    Reference values for HalfCheetah from Fu et al. 2020, Table 3:
        random baseline : -280.18
        expert baseline : 12135.0
    A score of 100 means the agent matched the expert; >100 means it exceeded it.
    """
    return (raw - d4rl_random) / (d4rl_expert - d4rl_random) * 100


print("display_video() and normalize_return() defined.")
````

## Section 5: Run 20 Rollouts & Save MP4s

This section:
1. Creates a MuJoCo `HalfCheetah-v4` environment with `render_mode='rgb_array'`
2. Rolls out the trained DT policy for 20 episodes
3. Saves each episode as an MP4 to `./videos/`
4. Collects episode returns for the summary in Section 7

> **Expected runtime on M5 Max:** approximately 3–6 minutes for 20 episodes at 1000 steps each. MuJoCo physics simulation is CPU-bound; the DT forward pass runs on MPS.

````python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 5: Run 20 rollout episodes and record each as an MP4
# ─────────────────────────────────────────────────────────────────────────────

# Create the MuJoCo environment in rgb_array render mode.
# This mode renders each frame silently to a numpy array — no OS window is opened,
# which avoids macOS event-loop conflicts with Jupyter.
env = gym.make(
    CFG.mujoco_env_id,
    render_mode='rgb_array',   # frames returned as (H, W, 3) numpy arrays
)

env.reset(seed=CFG.seed)

print(f"Environment created: {CFG.mujoco_env_id}")
print(f"Observation space  : {env.observation_space}")
print(f"Action space       : {env.action_space}")
print(f"\nRunning {CFG.n_episodes} episodes...")
print(f"  Target RTG  : {CFG.target_rtg:.0f}")
print(f"  Saving MP4s : {CFG.video_dir}/")
print()

# Storage for results
all_returns       = []
all_norm_returns  = []
all_video_paths   = []
all_reward_histories = []

for ep_idx in range(5):

    # Build the output path for this episode's MP4
    video_path = str(Path(CFG.video_dir) / f"episode_{ep_idx + 1:02d}_rtg{int(CFG.target_rtg)}.mp4")

    # Run the episode: physics simulation + DT policy inference + frame capture
    ep_return, ep_rewards = rollout_and_record(
        model        = model,
        env          = env,
        target_rtg   = CFG.target_rtg,
        state_mean   = STATE_MEAN,
        state_std    = STATE_STD,
        device       = DEVICE,
        context_len  = CFG.context_len,
        rtg_scale    = CFG.rtg_scale,
        video_path   = video_path,
        fps          = CFG.video_fps,
        max_ep_len   = CFG.max_ep_len,
    )

    # Compute D4RL-normalized score for this episode
    norm_ret = normalize_return(ep_return, CFG.d4rl_random, CFG.d4rl_expert)

    all_returns.append(ep_return)
    all_norm_returns.append(norm_ret)
    all_video_paths.append(video_path)
    all_reward_histories.append(ep_rewards)

    print(f"  Episode {ep_idx + 1:02d}/{CFG.n_episodes} "
          f"| Return: {ep_return:8.1f} "
          f"| D4RL norm: {norm_ret:5.1f} "
          f"| Steps: {len(ep_rewards):4d} "
          f"| Saved: {Path(video_path).name}")

# Close the environment — all frames are now written to disk
env.close()

print(f"\nAll {CFG.n_episodes} episodes complete.")
print(f"MP4 files saved to: {Path(CFG.video_dir).resolve()}")
````
